```python
import pandas as pd
import time

from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("Benchmark") \
    .getOrCreate()


# PANDAS

pandas_df = pd.read_csv(
    "data/raw/ipumsi_00001.csv.gz"
)

start = time.time()

pandas_result = pandas_df.groupby(
    'COUNTRY'
)['INCTOT'].mean()

end = time.time()

print(
    "Pandas:",
    end - start
)

# SPARK

spark_df = spark.read.csv(
    "data/raw/ipumsi_00001.csv.gz",
    header=True,
    inferSchema=True
)

start = time.time()

spark_result = spark_df.groupBy(
    'COUNTRY'
).mean('INCTOT')

spark_result.show()

end = time.time()

print(
    "Spark:",
    end - start
)
```
